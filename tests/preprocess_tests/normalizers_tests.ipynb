```python
from pathlib import Path
from ecg_analyzer.io import load_ecg_file
from ecg_analyzer.utils import plot_ecg
from ecg_analyzer.preprocess import z_score_normalization
```

```python
test_path = Path.cwd()
```

# Test 100

```python
rec = load_ecg_file(test_path, "00001_lr", "100")
plot_ecg(rec, title="raw signal")
```

```python
normalized_leads = z_score_normalization(rec.leads)
rec = rec.model_copy(update={"leads": normalized_leads})
plot_ecg(rec, title="normalized leads")
```
